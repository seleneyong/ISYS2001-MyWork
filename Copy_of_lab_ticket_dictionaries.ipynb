## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
# Build your dictionary from the transactions, and write your three functions here.
# Storing the same information with a dictionary
# The category names are the keys, and the total spent is the value
spending= {}
for transaction in transactions:
  category = transaction ['category']
  amount = transaction['amount']

  if category in spending:
    spending[category] +=amount
  else:
    spending[category]= amount

# using function total_for(spending, category) & .get() to look up the total amount spent in a category,
# and returns 0 for a category that was never recorded.

def total_for(spending, category):
  return spending.get(category,0)

# Using function add_transaction(spending, transaction) to add a new transaction
def add_transaction(spending,transaction):
    category= transaction['category']
    amount = transaction ['amount']

    if category in spending:
      spending[category] +=amount
    else:
      spending[category]= amount

# using function summary(spending) to summarise the spending, and .items() to go through the whole dictionary
def summary(spending):
  result={}

  for category, amount in spending.items():
    result[category] =amount

  return result
```

```python
# Your own feature goes here.
# Finding the category that spent the most

biggest=max(spending, key=spending.get)

print(f'Biggest category: {biggest}')
print(f'Amount spent: ${spending[biggest]}')

```
