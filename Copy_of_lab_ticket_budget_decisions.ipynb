## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Built a tool with the aim of helping users to make smart budgeting decision, it is required to ask the user what financial goal are they aiming, the estimated cost of the goal, when are they are expecting to achieve the goal(in years), and their monthly income and expenses.
 If the goal is to buy a item cost that is greater or equal to 1,000,000 and the expected time to reach is more than or equal to 5 years, then the tool should check wheter the monthly income is greater than monthly expenses. If yes, inform user they have a positive cash flow toward their goal. Else, recommend reducing expenses or increasing income before pursuing the goal.

```

````python
# Smart Budgeting Decision Tool

# Get user input
financial_goal = input("What financial goal are you aiming for? ")
estimated_cost = float(input("What is the estimated cost of your goal? "))
expected_time = float(input("In how many years do you expect to achieve this goal(in years)? "))
monthly_income = float(input("What is your monthly income? "))
monthly_expenses = float(input("What are your monthly expenses? "))

print("\n=== Budget Analysis ===")
print("Financial Goal:", financial_goal)

# Check the conditions
if estimated_cost >= 1000000 and expected_time >= 5:
    if monthly_income > monthly_expenses:
        print("\nYou have a positive cash flow toward your financial goal.")
    else:
        print("\nWe recommend reducing your expenses or increasing your income before pursuing this goal.")
else:
    print("\nYour financial goal does not meet the conditions for this assessment.")
````
