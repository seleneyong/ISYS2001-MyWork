## Part 1: Design before code

Before writing any loops, it is worth thinking through the patterns the problem needs.

**Scenario.** You want to record a series of expenses in one session, but you do not know in advance how many there will be.

A reasonable plan, written in plain English, might look like this:


```python
# One way to think through the design before writing code:
#
# 1. Ask the user for an expense, or a signal that they are finished.
# 2. If they signal they are finished, stop asking.
# 3. Otherwise, check the input is a valid positive number.
#    - if it is not, explain the problem and ask again.
# 4. If it is valid, store it and add it to a running total.
# 5. Report the running total, then return to step 1.
#
# Notice there are two different kinds of repetition here:
#   - "keep asking until finished" has no fixed count  -> while loop
#   - "check until the input is valid" also has no fixed count -> while loop
```

## Part 2: While loops for validation

The most common use of a while loop is to keep asking until the input is acceptable. The pattern is:

1. start with a value that is deliberately invalid
2. loop while the value is invalid
3. read input inside the loop
4. check it, and give feedback
5. the loop ends once a valid value is entered


```python
print("=== Expense entry with validation ===")

expense = -1  # start with a deliberately invalid value

while expense < 0:
    try:
        expense = float(input("Enter expense amount: $"))
        if expense < 0:
            print("An expense cannot be negative. Please try again.")
    except ValueError:
        print("Please enter a valid number.")
        expense = -1  # keep the loop going

print(f"Valid expense recorded: ${expense:.2f}")
```

Here is the same pattern applied to a monthly budget, where a valid value sits between $100 and $50,000. Read how the loop condition captures both limits.


```python
print("=== Monthly budget entry ===")

budget = 0  # invalid to begin with

while budget < 100 or budget > 50000:
    try:
        budget = float(input("Enter monthly budget ($100 to $50,000): $"))
        if budget < 100:
            print("Budget too low. Minimum is $100.")
        elif budget > 50000:
            print("Budget too high. Maximum is $50,000.")
    except ValueError:
        print("Please enter a valid number.")
        budget = 0

print(f"Budget set to: ${budget:.2f}")
```

## Part 3: Sentinel values

Sometimes you do not know how many times to repeat. A sentinel value is a special input that means "stop". Here the sentinel is the word `done`.


```python
print("=== Multiple expense entry ===")

expenses = []
total = 0

print("Enter your expenses. Type 'done' when finished.")

while True:
    entry = input("Expense amount (or 'done'): $")

    if entry.lower() == "done":
        break  # the sentinel: leave the loop

    try:
        expense = float(entry)
        if expense >= 0:
            expenses.append(expense)
            total += expense
            print(f"  Added ${expense:.2f}. Running total: ${total:.2f}")
        else:
            print("  An expense must be positive.")
    except ValueError:
        print("  Please enter a valid number, or 'done'.")

print(f"\nSummary: {len(expenses)} expenses totalling ${total:.2f}")
```

The same idea works for text. Here we collect category names until the user types `finish`, and we reject empty input along the way.


```python
print("=== Expense categories ===")

categories = []

print("Enter expense categories. Type 'finish' when done.")

while True:
    category = input("Category name (or 'finish'): ").strip()

    if category.lower() == "finish":
        break

    if category == "":
        print("  A category cannot be empty.")
        continue

    categories.append(category)
    print(f"  Added '{category}'. You now have {len(categories)} categories.")

print(f"\nCategories recorded: {categories}")
```

## Part 4: For loops for a known count

When you know exactly how many steps you need, a for loop is the natural fit. Below we plan a spend for each workday. Note the validation while loop nested inside the for loop.


```python
print("=== Weekly expense planner ===")

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
weekly_expenses = []
total_week = 0

for day in days:
    while True:  # validation loop inside the for loop
        try:
            expense = float(input(f"{day} planned expense: $"))
            if expense >= 0:
                weekly_expenses.append(expense)
                total_week += expense
                break  # valid, leave the validation loop
            else:
                print("  The amount cannot be negative.")
        except ValueError:
            print("  Please enter a valid number.")

print(f"\nWeekly plan: ${total_week:.2f} total")
print(f"Daily average: ${total_week / len(days):.2f}")
```

Here is a savings goals example using `range(3)` for a fixed number of goals, storing each goal's details and reporting progress.


```python
print("=== Savings goals ===")

goal_names = []
goal_targets = []
goal_current = []

for i in range(3):
    print(f"\nGoal {i + 1}")
    name = input("  Goal name: ")
    target = float(input("  Target amount: $"))
    current = float(input("  Saved so far: $"))

    goal_names.append(name)
    goal_targets.append(target)
    goal_current.append(current)

print("\n--- Progress ---")
for i in range(len(goal_names)):
    percent = (goal_current[i] / goal_targets[i]) * 100 if goal_targets[i] > 0 else 0
    remaining = goal_targets[i] - goal_current[i]
    print(f"{goal_names[i]}: {percent:.1f}% complete, ${remaining:.2f} to go")
```

## Part 5: Putting it together

The patterns above combine into a small program: set up the user, collect expenses with a sentinel, then summarise. This is the shape you will rebuild in your own words in the mini-project.


```python
print("=== Finance tracker ===")

name = input("Enter your name: ")

# setup: budget validation
budget = 0
while budget < 100 or budget > 50000:
    try:
        budget = float(input("Monthly budget ($100 to $50,000): $"))
        if budget < 100:
            print("Budget too low. Minimum is $100.")
        elif budget > 50000:
            print("Budget too high. Maximum is $50,000.")
    except ValueError:
        print("Please enter a valid number.")
        budget = 0

print(f"\nWelcome {name}. Budget: ${budget:.2f}")

# collection: expenses with a sentinel
expenses = []
total_spent = 0
print("\nEnter your expenses. Type 'done' when finished.")
while True:
    entry = input("Expense (or 'done'): $")
    if entry.lower() == "done":
        break
    try:
        amount = float(entry)
        if amount >= 0:
            expenses.append(amount)
            total_spent += amount
            print(f"  Added ${amount:.2f}. Total: ${total_spent:.2f}")
        else:
            print("  The amount must be positive.")
    except ValueError:
        print("  Please enter a valid number, or 'done'.")

# summary
if len(expenses) > 0:
    large_count = 0
    for amount in expenses:
        if amount > budget * 0.1:  # more than 10% of budget
            large_count += 1

    print(f"\nTotal spent: ${total_spent:.2f}")
    print(f"Budget used: {(total_spent / budget) * 100:.1f}%")
    print(f"Large expenses (over 10% of budget): {large_count}")
else:
    print("\nNo expenses recorded.")
```

## Part 6: Testing and debugging

Loops fail in a small number of predictable ways. When one misbehaves, work through this checklist:

- **Infinite loop?** Check that the condition can eventually become false.
- **Never runs?** Check the starting value against the condition.
- **Wrong number of steps?** Trace it by hand for a small input.
- **Validation not catching bad input?** Check the condition logic.

The cell below is a quick sanity check that your loops run at all.


```python
print("This should count 1 to 5:")

count = 1
while count <= 5:
    print(count)
    count = count + 1  # without this line the loop never ends

print("If you saw 1 to 5, your loops are working.")
```
